```python
df.drop_duplicates(inplace=True)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['popularity'],
    bins=30,
    kde=True
)

plt.title("Distribution of Spotify Track Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Tracks")

plt.show()
```

```python
top_genres = df['track_genre'].value_counts().head(10)


plt.figure(figsize=(10,5))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Genres Based on Number of Tracks")
plt.xlabel("Number of Songs")
plt.ylabel("Genre")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="energy",
    y="danceability",
    alpha=0.5
)

plt.title("Relationship Between Energy and Danceability")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    data=df,
    x="explicit",
    y="popularity"
)

plt.title("Popularity Comparison Between Explicit and Clean Songs")

plt.show()
```

```python
plt.figure(figsize=(12,8))

numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    df['tempo'],
    fill=True
)

plt.title("Distribution of Song Tempo")

plt.xlabel("Tempo")

plt.show()
```
